```python
from abc import ABC, abstractmethod
```

```python
class car(ABC):
    @ abstractmethod
    def airbag(self):
        pass
```

```python
class nexon(car):
    def airbag(self):
        return 'Air bag in nexon'
    def sunroof(self):
        return 'Sunroof in nexon'
```

```python
obj=nexon()
```

```python
obj.airbag()
```

```python
obj.sunroof()cl
```

```python
class fortuner(car):
    def button(self):
        return 'Button start'
```

```python
obj2=fortuner()
```

```python
class phone():
    @ abstractmethod
    def call(self):
        pass
```

```python
class moto(phone):
    def call(self):
        return 'call is having in moto'
    def screentouch(self):
        return 'screentouch'
```

```python
obj3=moto()
```

```python
obj3.call()
```

```python
obj3.screentouch()
```
