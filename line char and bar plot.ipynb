```python
# order_date vs Gross_Amount
plt.figure(figsize=(10,5))
plt.plot(df["order_date"], df["gross_amount"],
         label="Gross Amount")
plt.title("Date-wise Gross Sales")
plt.xlabel("Order Date")
plt.ylabel("Gross Amount")
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.show()
```

```python
# 2 order_date vs net amount
plt.figure(figsize=(10, 5))
plt.plot(df["order_date"], df["net_amount"],
         label="Net Amount", marker="o")
plt.title("Date-wise Net Sales")
plt.xlabel("Order Date")
plt.ylabel("Net Amount")
plt.legend()
plt.grid()
plt.show()
```

```python
# 3 Quantity vs Unti Price
plt.figure(figsize=(10,5))
plt.plot(df["quantity"], df["unit_price"],
         label="Quantity vs Unit Price")
plt.title("Quantity vs Unit Price")
plt.xlabel("Quantity")
plt.ylabel("Unit Price")
plt.legend()
plt.grid()
plt.show()
```

```python
# 4 Unit Price vs Gross amount 
plt.figure(figsize=(10,5))
plt.plot(df["unit_price"], df["gross_amount"],
         label="Unit Price vs Gross Amount")
plt.title("Unit Price vs Gross Amount")
plt.xlabel("Unit Price")
plt.ylabel("Gross Amount")
plt.legend()
plt.grid()
plt.show()
```

```python
# 5 Costomer_age vs Coustomer_ratting
plt.figure(figsize=(10,5))
plt.plot(df["customer_age"], df["customer_rating"],
         label="Customer Age vs Customer Rating")
plt.title("Customer Age vs Customer Rating")
plt.xlabel("Customer Age")
plt.ylabel("Customer Rating")
plt.legend()
plt.grid()
plt.show()
```

```python
# 1 Product Category vs net ammount
data = df.groupby("product_category")["net_amount"].sum() # Calculates the total net amount for each category.
plt.figure(figsize=(10,5))
data.plot(kind="bar")
plt.title("Product Category vs net ammount")
plt.xlabel("Product Category")
plt.ylabel("net ammount")
plt.xticks(rotation=45)
plt.grid()
plt.legend()
plt.show()
```

```python
# 2 Region vs net amount
region_sales = df.groupby("region")["net_amount"].sum()
plt.figure(figsize=(8,5))
region_sales.plot(kind="bar")
plt.title("Region vs Total Sales")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.grid()
plt.legend()
plt.show()
```

```python
# 3 Sales Channel vs Total Sales
channel_sales = df.groupby("sales_channel")["net_amount"].sum()
plt.figure(figsize=(8,5))
channel_sales.plot(kind="bar")
plt.title("Sales Channel vs Total Sales")
plt.xlabel("Sales Channel")
plt.ylabel("Total Sales")
plt.grid()
plt.legend()
plt.show()
```

```python
# Product Category vs Quantity
plt.figure(figsize=(9,5))
df.groupby("product_category")["quantity"].sum().plot(kind="bar")
plt.title("Product Category vs Quantity")
plt.xlabel("Product Category")
plt.ylabel("Quantity")
plt.xticks(rotation=45)
plt.grid()
plt.legend()
plt.show()
```

```python
# 5 Shipping Type vs Delivery Days
plt.figure(figsize=(8,5))
df.groupby("shipping_type")["delivery_days"].mean().plot(kind="bar") # Calculates the average delivery days for each shipping type.
plt.title("Shipping Type vs  Delivery Days")
plt.xlabel("Shipping Type")
plt.ylabel("Delivery Days")
plt.grid()
plt.legend()
plt.show()
```
