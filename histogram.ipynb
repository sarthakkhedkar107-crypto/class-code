```python
import pandas as pd
import matplotlib.pyplot as plt
```

# home work 

```python
# 1 create Histrogram of Coustmer age 

plt.figure(figsize=(8,5))
plt.hist(df["customer_age"], bins=10, edgecolor="black")
plt.title("Distribution of Customer Age")
plt.xlabel("Customer Age")
plt.ylabel("Number of Customers")
plt.grid()
plt.show()
```

```python
# 2 create a histrogram of qunatity
plt.figure(figsize=(8,5))
plt.hist(df["quantity"], bins=10, edgecolor="black")
plt.title("Distribution of Product Quantity")
plt.xlabel("Quantity")
plt.ylabel("Number of Orders")
plt.grid()
plt.show()
```

```python
# 3 create histrogram of net_amount
plt.figure(figsize=(8,5))
plt.hist(df["net_amount"], bins=10, edgecolor="black")
plt.title("Distribution of Net Amount")
plt.xlabel("Net Amount")
plt.ylabel("Number of Orders")
plt.grid(axis="y")
plt.show()
```

```python
 # 4 histogram bins 5
plt.figure(figsize=(15,5))
plt.hist(df["customer_age"], bins=5, edgecolor="black")
plt.title("Customer Age - 5 Bins")
plt.xlabel("Customer Age")
plt.ylabel("Number of Customers")
plt.grid()
plt.show()

```

```python
# Histogram with 10 bins
plt.hist(df["customer_age"], bins=10, edgecolor="black")
plt.title("Customer Age - 10 Bins")
plt.xlabel("Customer Age")
plt.ylabel("Number of Customers")
plt.grid(axis="y")
plt.show()
```

```python
# Histogram with 20 bins
plt.hist(df["customer_age"], bins=20, edgecolor="black")
plt.title("Customer Age - 20 Bins")
plt.xlabel("Customer Age")
plt.ylabel("Number of Customers")
plt.grid()
plt.show()
```

```python
# 5 display mean using plt.axvline
mean_age = df["customer_age"].mean()
plt.figure(figsize=(8,5))
plt.hist(df["customer_age"].dropna(), bins=10, edgecolor="black")
plt.axvline(mean_age, color="red", linestyle="--",
            linewidth=2, label="Mean")
plt.title("Customer Age Histogram with Mean")
plt.xlabel("Customer Age")
plt.ylabel("Number of Customers")
plt.legend()
plt.grid()
plt.show()
```

```python
# display  medain using plt.axvline
median_age = df["customer_age"].median()
plt.figure(figsize=(8,5))
plt.hist(df["customer_age"].dropna(), bins=10, edgecolor="black")
plt.axvline(median_age, color="green", linestyle="--",
            linewidth=2, label="Median")
plt.title("Customer Age Histogram with Median")
plt.xlabel("Customer Age")
plt.ylabel("Number of Customers")
plt.legend()
plt.grid()
plt.show()
```
