```python
import pandas as pd
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/refs/heads/master/titanic.csv')
df
```

```python
avg=df['Fare'].mean()
avg
```

```python
df['Fare']>=avg
```

```python
df[df['Fare']>=avg]['Name']
```

```python
def cat(n):
    if n>250:
        return 'High'
    elif n>100:
        return 'Medium'
    else:
        return 'Low'
        
```

```python
df['fare_cat']=df['Fare'].apply(cat)
```

```python
df
```

```python
df.isnull().sum()
```

```python

age_mean=df['Age'].mean()
```

```python
age_mean
```

```python
df['Age']=df['Age'].fillna(age_mean)
```
