```python
import logging
logging.basicConfig(filename='project1.log',level = logging.INFO, 
              format= "%(asctime)s - %(levelname)s - %(message)s" )
```

```python
def div(a,b):
    logging.info(f'value of a{a} value of b{b}')
    try:
        result=a/b
        logging.info(f'value of a{a} value of b{b} result is {result}')
        logging.info('Project complited')
        return result
    except  Exception  as e:
        logging.error('The zero division error')
        logging.error(e)
```

```python
import logging

logging.basicConfig(filename="error.log", level=logging.ERROR)

def read_file():
    try:
        f = open("data.txt", "r")
        print(f.read())
        f.close()
    except Exception as e:
        logging.error(e)
        print("Error:", e)

read_file()
```

```python
import logging

logging.basicConfig(filename="error.log", level=logging.ERROR)

def show_item():
    try:
        lst = [10, 20, 30]
        i = int(input("Enter index: "))
        print(lst[i])
    except Exception as e:
        logging.error(e)
        print("Error:", e)

show_item()
```
