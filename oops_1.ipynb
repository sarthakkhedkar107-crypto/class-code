```python
class bank():
    def __init__(self,balance):
        self.balance=balance
    def intrest(self):
        return self.balance*0.10
    def fd(self,months):
        total = self.balance * (0.10*months)
        return total
            
```

# Inheritance

```python
class parent():
    def singing(self):
        return ' parent Good in singing'
```

```python
class child(parent):
    def cricket(self):
        return 'i am Good in cricket'
```

```python
obj4=parent()
```

```python
obj4.singing()
```

```python
obj5=child()
```

```python
obj5.singing()
```

```python
obj5.cricket()
```

```python
class car():
    def __init__(self,engine,fuel,seat,color): # constructor way to pass data to class
        self.engine = engine
        self.fuel = fuel
        self.seat = seat
        self.color = color
    def start(self):
        return 'Car is starting'
```

```python
class bus(car):
    def off(self):
        return ' bus geting off '
```

```python
obj7 = bus(500,'Diesel',30,'White')
```

```python
obj7.start()
```

```python
obj7.off()
```

```python
obj7.fuel
```
