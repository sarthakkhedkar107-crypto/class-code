```python
import pandas as pd 
```

```python
df = pd.read_csv("C:/Users/Vinod/Downloads/industry_retail_orders_dirty_1000.csv")
print(df)
```

```python
df.info()
```

```python
# coustmer age

df["customer_age"] = df["customer_age"].fillna(
    df["customer_age"].mean()
)
df
```

```python
#customer_segment 
df["customer_segment"] = df["customer_segment"].fillna(
    df["customer_segment"].mode()[0]
)
df
```

```python
# city
df["city"]=df["city"].fillna(
    df["city"].mode()[0]
)
df
```

```python
#   product_category    
df["product_category"]=df["product_category"].fillna(
    df["product_category"].mode()[0]
)
df
```

```python
df["unit_price"]=df["unit_price"].fillna(
    df["unit_price"].mean()
)
df
```

```python
# discount_pct
df["discount_pct"]=df["discount_pct"].fillna(
    df["discount_pct"].median()
)
df
```

```python
# shipping_fee
df["shipping_fee"]=df["shipping_fee"].fillna(
    df["shipping_fee"].mean()
)
df
```

```python
# delivery_days 
df["delivery_days"]=df["delivery_days"].fillna(
    df["delivery_days"].mean()
)
df
```

```python
# payment_method   
df["payment_method"]=df["payment_method"].fillna(
    df["payment_method"].mode()[0]
)
df
```

```python
#  actual_delivery_date 
df["actual_delivery_date"]=df["actual_delivery_date"].fillna(
    df["actual_delivery_date"].mode()[0]
)
df
```

```python
# marketing_campaign
df["marketing_campaign"]=df["marketing_campaign"].fillna(
    df["marketing_campaign"].mode()[0]
)
df
```

```python
# customer_rating    
df["customer_rating"]=df["customer_rating"].fillna(
    df["customer_rating"].mode()[0]
)
df
```
